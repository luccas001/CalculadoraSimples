```python
 # CALCULADORA

 #SOMA

def somar (a, b): #def seria a função -
  somar = a + b
  return print(f" A soma dos números {a} + {b} é igual a {somar} ")

def subtrair (a, b):
  subtrair = a - b
  return print(f" A subtração dos númerios {a} + {b} é igual a {subtrair}")

def multiplicar (a, b):
  multiplicar = a * b
  return print (f" A multiplicação dos números {a} x {b} é igual a {multiplicar}")

def dividir (a, b):
  if b == 0:
    return print("Erro: Não é possível dividir um número por zero")
  else:
    dividir = a / b
    return print(f"A divisão dos números {a} / {b} é igual a {dividir}")

while True:
    print("\n --- CALCULADORA PYTHON ---")
    print("1 - SOMAR")
    print("2 - SUBTRAIR")
    print("3 - MULTIPLICAR")
    print("4 - DIVIDIR")
    print("0 - SAIR")

    escolha = input("Escolha uma opção ")

    if escolha == "0":
      print(" Calculadora encerrada")
      break

    if escolha in ["1","2","3","4"]: #Números entre "" porque ainda se mantém como strings
      num1 = float(input(" Digite o primeiro número: "))
      num2 = float(input(" Digite o segundo número: "))

      if escolha == "1":
        resultado = somar(num1, num2)
      elif escolha == "2":
        resultado = subtrair(num1, num2)
      elif escolha == "3":
        resultado = multiplicar(num1, num2)
      elif escolha == "4":
        resultado = dividir (num1, num2)

    else:
      print(" - Escolha uma opção válida -  ")










```
